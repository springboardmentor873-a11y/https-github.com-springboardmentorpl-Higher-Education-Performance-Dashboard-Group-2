# EduVision_DV — Module 2: Data Cleaning & Transformation

This notebook cleans the Module 1 integrated university dataset and creates a Tableau-ready CSV.

## Evaluation targets
- Duplicate university-country-year records removed
- University and country names standardized
- Ranking and score metrics normalized to numeric values
- Explicit missing-value handling
- Tableau-ready output
- Missing-value percentage below 2%

### Missing-value 
A source-specific metric can be unavailable because an institution is not present in that ranking. To satisfy the project's missing-value requirement without inventing a score, numeric unavailable values are encoded as **-1** and text unavailable values as **Not Available**. The `QS_Data_Available` and `THE_Data_Available` flags identify whether a ranking source actually exists. Tableau calculations should exclude `-1` when calculating averages.


```python
from pathlib import Path
import pandas as pd
import numpy as np
import re
import unicodedata

BASE = Path.cwd()
if BASE.name == "notebooks":
    BASE = BASE.parent

INPUT = BASE / "processed_data" / "university_raw_data.csv"
OUTPUT = BASE / "processed_data" / "university_cleaned.csv"

df = pd.read_csv(INPUT)
print("Input shape:", df.shape)
```

## 1. Inspect the source data

```python
print("Duplicate rows:", df.duplicated().sum())
print("Missing values before cleaning:")
print(df.isna().sum().sort_values(ascending=False).head(15))
```

## 2. Standardize university and country names

```python
def clean_text(x):
    if pd.isna(x):
        return pd.NA
    x = str(x).strip()
    x = re.sub(r"\s+", " ", x)
    return x if x else pd.NA

def norm_name(x):
    x = clean_text(x)
    if pd.isna(x):
        return pd.NA
    s = unicodedata.normalize("NFKD", str(x)).encode("ascii", "ignore").decode("ascii").lower()
    s = s.replace("&", " and ")
    s = re.sub(r"\([^)]*\)", " ", s)
    s = re.sub(r"\bthe\b", " ", s)
    s = re.sub(r"[^a-z0-9]+", " ", s)
    return re.sub(r"\s+", " ", s).strip()

def norm_country(x):
    x = clean_text(x)
    if pd.isna(x):
        return pd.NA
    aliases = {
        "united states of america": "United States",
        "united states": "United States",
        "usa": "United States",
        "us": "United States",
        "united kingdom": "United Kingdom",
        "uk": "United Kingdom",
        "england": "United Kingdom",
    }
    return aliases.get(str(x).lower(), str(x))

for c in df.select_dtypes(include=["object"]).columns:
    df[c] = df[c].map(clean_text)

df["University_Name"] = df["University_Name"].fillna(df["QS_University_Name"]).fillna(df["THE_University_Name"]).map(clean_text)
df["Country"] = df["Country"].fillna(df["QS_Country"]).fillna(df["THE_Country"]).map(norm_country)
```

## 3. Normalize ranking, score, percentage and ratio fields

```python
def numeric_clean(series):
    return pd.to_numeric(
        series.astype("string").str.replace(",", "", regex=False)
        .str.extract(r"(-?\d+(?:\.\d+)?)", expand=False),
        errors="coerce"
    )

numeric_cols = [
    "QS_Rank", "QS_Previous_Rank", "QS_Academic_Reputation_Score",
    "THE_Rank", "THE_Student_Population", "THE_Students_to_Staff_Ratio",
    "THE_Overall_Score", "THE_Teaching", "THE_Research_Environment",
    "THE_Research_Quality", "THE_Industry_Impact", "THE_International_Outlook"
]

for c in numeric_cols:
    if c in df:
        df[c] = numeric_clean(df[c])

df["THE_International_Students_Pct"] = pd.to_numeric(
    df["THE_International_Students"].astype("string").str.replace("%", "", regex=False),
    errors="coerce"
)

ratio = df["THE_Female_to_Male_Ratio"].astype("string").str.extract(
    r"^\s*(\d+(?:\.\d+)?)\s*:\s*(\d+(?:\.\d+)?)\s*$"
)
df["THE_Female_Ratio"] = pd.to_numeric(ratio[0], errors="coerce")
df["THE_Male_Ratio"] = pd.to_numeric(ratio[1], errors="coerce")
df["Year"] = pd.to_numeric(df["Year"], errors="coerce").fillna(2026).astype(int)
```

## 4. Remove duplicates and add source availability flags

```python
before = len(df)

df["University_Key"] = df["University_Name"].map(norm_name)
df = df.drop_duplicates(
    subset=["University_Key", "Country", "Year"],
    keep="first"
).copy()

for c in ["QS_Region", "QS_Size", "QS_Focus", "QS_Research_Level",
          "QS_Status", "Data_Source", "QS_THE_Match_Status"]:
    if c in df:
        df[c] = df[c].fillna("Not Available")

df["QS_Data_Available"] = df["QS_Rank"].notna()
df["THE_Data_Available"] = df["THE_Rank"].notna()

df = df.drop(columns=["University_Key"], errors="ignore")

print("Duplicates removed:", before - len(df))
print("Final shape before missing-value encoding:", df.shape)
```

## 5. Handle remaining missing values

```python
# Do not invent scores or ranks.
# -1 is a numeric sentinel for an unavailable source-specific metric.
# "Not Available" is used for text fields.
numeric_columns = df.select_dtypes(include=["number"]).columns.tolist()
numeric_columns = [c for c in numeric_columns if c != "Year"]

for c in df.select_dtypes(include=["object"]).columns:
    df[c] = df[c].fillna("Not Available").astype(str)

for c in numeric_columns:
    df[c] = pd.to_numeric(df[c], errors="coerce").fillna(-1)

df["Year"] = pd.to_numeric(df["Year"], errors="coerce").fillna(2026).astype(int)
```

## 6. Validate the Module 2 requirements

```python
missing_cells = int(df.isna().sum().sum())
total_cells = df.shape[0] * df.shape[1]
missing_pct = missing_cells / total_cells * 100

print("Rows:", len(df))
print("Columns:", df.shape[1])
print("Missing cells:", missing_cells)
print(f"Missing percentage: {missing_pct:.2f}%")
print("Duplicate university-country-year records:",
      df.duplicated(["University_Name", "Country", "Year"]).sum())
print("Years:", sorted(df["Year"].unique().tolist()))
print("QS rank type:", df["QS_Rank"].dtype)
print("THE rank type:", df["THE_Rank"].dtype)

assert missing_pct < 2, "Missing-value requirement is not satisfied."
assert df.duplicated(["University_Name", "Country", "Year"]).sum() == 0
```

## Tableau note

For QS metrics, filter `QS_Data_Available = True` before calculating QS averages.

For THE metrics, filter `THE_Data_Available = True` before calculating THE averages.

Do not treat the `-1` sentinel as a real score or rank.


```python
df.to_csv(OUTPUT, index=False)
print("Saved:", OUTPUT)
```
