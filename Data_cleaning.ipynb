```python
# Handling Missing Values in QS Ranking Dataset
# Import required libraries

import pandas as pd
import numpy as np

from sklearn.impute import SimpleImputer

# 1. Load Dataset

df = pd.read_csv("/content/2026_QS_World University_Rankings.csv")


# Display first 5 rows

print("First 5 rows of dataset:")
print(df.head())

# 2. Dataset Information

print("\nDataset Information:")
print(df.info())


# Shape of dataset

print("\nDataset Shape:")
print(df.shape)



# ==========================================
# 3. Check Missing Values
# ==========================================

print("\nMissing Values in Each Column:")
print(df.isnull().sum())


print("\nTotal Missing Values:")
print(df.isnull().sum().sum())



# ==========================================
# 4. Separate Numerical and Categorical Columns
# ==========================================

numerical_cols = df.select_dtypes(
    include=['int64','float64']
).columns


categorical_cols = df.select_dtypes(
    include=['object']
).columns



print("\nNumerical Columns:")
print(list(numerical_cols))


print("\nCategorical Columns:")
print(list(categorical_cols))



# ==========================================
# 5. Handle Numerical Missing Values
#    Replace with Mean
# ==========================================

num_imputer = SimpleImputer(strategy='mean')


df[numerical_cols] = num_imputer.fit_transform(
    df[numerical_cols]
)



# ==========================================
# 6. Handle Categorical Missing Values
#    Replace with Mode
# ==========================================

cat_imputer = SimpleImputer(strategy='most_frequent')


df[categorical_cols] = cat_imputer.fit_transform(
    df[categorical_cols]
)



# ==========================================
# 7. Check Missing Values After Cleaning
# ==========================================

print("\nMissing Values After Handling:")
print(df.isnull().sum())


print("\nTotal Missing Values After Cleaning:")
print(df.isnull().sum().sum())



# ==========================================
# 8. Display Cleaned Dataset
# ==========================================

print("\nCleaned Dataset:")
print(df.head())



# ==========================================
# 9. Save Clean Dataset
# ==========================================

df.to_csv(
    "QS_Ranking_Cleaned_Dataset.csv",
    index=False
)


print("\nDataset cleaned and saved successfully!")
```

```python
# =====================================================
# Missing Value Handling - World University Rankings 2026
# =====================================================


# Import Libraries

import pandas as pd
import numpy as np

from sklearn.impute import SimpleImputer



# =====================================================
# 1. Load Dataset
# =====================================================

df = pd.read_csv("/content/world_university_rankings_2026.csv")


# Display first rows

print("First 5 rows:")
print(df.head())



# =====================================================
# 2. Dataset Information
# =====================================================

print("\nDataset Information:")
df.info()


print("\nDataset Shape:")
print(df.shape)



# =====================================================
# 3. Check Missing Values
# =====================================================

print("\nMissing Values in Each Column:")
print(df.isnull().sum())


print("\nTotal Missing Values:")
print(df.isnull().sum().sum())



# =====================================================
# 4. Separate Column Types
# =====================================================


# Numerical columns

numerical_cols = df.select_dtypes(
    include=['int64','float64']
).columns


# Categorical columns

categorical_cols = df.select_dtypes(
    include=['object']
).columns


# Boolean columns

boolean_cols = df.select_dtypes(
    include=['bool']
).columns



print("\nNumerical Columns:")
print(list(numerical_cols))


print("\nCategorical Columns:")
print(list(categorical_cols))


print("\nBoolean Columns:")
print(list(boolean_cols))



# =====================================================
# 5. Handle Numerical Missing Values
#    Using Mean
# =====================================================

num_imputer = SimpleImputer(
    strategy="mean"
)


df[numerical_cols] = num_imputer.fit_transform(
    df[numerical_cols]
)



# =====================================================
# 6. Handle Categorical Missing Values
#    Using Mode
# =====================================================

cat_imputer = SimpleImputer(
    strategy="most_frequent"
)


df[categorical_cols] = cat_imputer.fit_transform(
    df[categorical_cols]
)



# =====================================================
# 7. Handle Boolean Missing Values
#    Using Mode
# =====================================================
# Based on the df.isnull().sum() output, the boolean column 'ranked_in_all_3' has no missing values.
# Therefore, this step is not needed. If there were missing boolean values, consider converting to int (0/1) or object type first.
# bool_imputer = SimpleImputer(
#     strategy="most_frequent"
# )


# df[boolean_cols] = bool_imputer.fit_transform(
#     df[boolean_cols]
# )



# =====================================================
# 8. Verify Missing Values
# =====================================================

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())


print("\nTotal Missing Values After Cleaning:")
print(df.isnull().sum().sum())



# =====================================================
# 9. Display Clean Dataset
# =====================================================

print("\nCleaned Dataset:")
print(df.head())



# =====================================================
# 10. Save Cleaned Dataset
# =====================================================

df.to_csv(
    "world_university_rankings_2026_cleaned.csv",
    index=False
)


print("\nCleaned dataset saved successfully!")
```
