```python
import pandas as pd
import numpy as np

# Load Excel file
df = pd.read_excel("/content/merged_dataset (1).xlsx")

# -------------------------------
# Data Cleaning
# -------------------------------

# Standardize column names
df.columns = (
    df.columns.str.strip()
              .str.lower()
              .str.replace(" ", "_")
              .str.replace("/", "_")
)

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Remove completely empty rows and columns
df.dropna(how="all", inplace=True)
df.dropna(axis=1, how="all", inplace=True)

# Remove extra spaces from text columns
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].astype(str).str.strip()

# Replace invalid values with NaN
df.replace(["", " ", "NA", "N/A", "null", "None"], np.nan, inplace=True)

# Fill missing numeric values with median
for col in df.select_dtypes(include=["number"]).columns:
    df[col] = df[col].fillna(df[col].median())

# Fill missing text values with "Unknown"
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna("Unknown")

# Merge qs_rank and the_rank into rank
if "qs_rank" in df.columns and "the_rank" in df.columns:
    df["rank"] = df["qs_rank"].combine_first(df["the_rank"])

# Remove duplicate universities
if "university_name" in df.columns:
    df.drop_duplicates(subset=["university_name"], inplace=True)

# Reset index
df.reset_index(drop=True, inplace=True)

# Save cleaned Excel file
df.to_excel("university_cleaned_data.xlsx", index=False)

print("Data cleaned successfully!")
print("Final Shape:", df.shape)
print("Saved as: university_cleaned_data.xlsx")

```
