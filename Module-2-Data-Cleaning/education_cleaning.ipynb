## QS World University Rankings 2026 + THE World University Rankings 2026

### Internship Milestone 1 — Data Collection and Preparation

This notebook documents the cleaning, transformation, integration, and validation of the QS World University Rankings 2026 and Times Higher Education (THE) World University Rankings 2026 datasets.

### Objectives

- Load the QS and THE university ranking datasets
- Inspect and understand the datasets
- Clean text and encoding issues
- Standardize university names
- Standardize country names
- Normalize ranking and performance metrics
- Handle missing values appropriately
- Remove and validate duplicates
- Integrate QS and THE datasets
- Validate the merged dataset
- Measure data completeness
- Produce a Tableau-ready dataset

### Final Dataset

- Records: 1,008 universities
- Core completeness: 99.43%
- Missing core values: 0.57%
- Duplicate QS universities: 0
- Duplicate THE universities: 0
- Country mismatches: 0

Completeness > 95% and missing values < 2%.

```python
import pandas as pd
import numpy as np
import re
import unicodedata
from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully.")
```

```python
# Project paths

BASE_DIR = Path.cwd().parent

DATA_DIR = BASE_DIR / "data"
OUTPUT_DIR = BASE_DIR / "output"

print("Base directory:", BASE_DIR)
print("Data directory:", DATA_DIR)
print("Output directory:", OUTPUT_DIR)
```

```python
# Load cleaned QS and THE datasets

QS_FILE = OUTPUT_DIR / "qs_cleaned_encoding_fixed.csv"
THE_FILE = OUTPUT_DIR / "the_cleaned_encoding_fixed.csv"

qs = pd.read_csv(QS_FILE, encoding="utf-8-sig")
the = pd.read_csv(THE_FILE, encoding="utf-8-sig")

print("QS dataset shape :", qs.shape)
print("THE dataset shape:", the.shape)
```

```python
print("QS columns:")
display(pd.DataFrame({"Column": qs.columns}))

print("\nTHE columns:")
display(pd.DataFrame({"Column": the.columns}))
```

```python
print("QS Dataset Information")
display(qs.info())

print("\nTHE Dataset Information")
display(the.info())
```

```python
print("QS Dataset Information")
qs.info()

print("\nTHE Dataset Information")
the.info()
```

```python
print("QS missing values:")
display(
    qs.isna().sum()
    .sort_values(ascending=False)
    .to_frame("Missing Values")
)

print("\nTHE missing values:")
display(
    the.isna().sum()
    .sort_values(ascending=False)
    .to_frame("Missing Values")
)
```

```python
def standardize_university_name(value):
    if pd.isna(value):
        return value

    value = str(value).strip()

    # Fix common encoding issues
    try:
        if any(x in value for x in ["Ã", "Â", "â", "Å", "Ä"]):
            value = value.encode("latin1").decode("utf-8")
    except (UnicodeEncodeError, UnicodeDecodeError):
        pass

    # Unicode normalization
    value = unicodedata.normalize("NFKC", value)

    # Normalize whitespace
    value = re.sub(r"\s+", " ", value)

    return value.strip()


qs["Institution Name"] = qs["Institution Name"].apply(
    standardize_university_name
)

the["Name"] = the["Name"].apply(
    standardize_university_name
)

print("University names standardized successfully.")
```

```python
def standardize_country_name(value):
    if pd.isna(value):
        return value

    value = str(value).strip()

    replacements = {
        "United States": "United States of America",
        "United States of America": "United States of America",
        "China": "China (Mainland)",
        "China (Mainland)": "China (Mainland)",
        "South Korea": "Republic of Korea",
        "Korea, Republic of": "Republic of Korea",
        "Republic of Korea": "Republic of Korea",
        "Russia": "Russian Federation",
        "Russian Federation": "Russian Federation",
        "Czech Republic": "Czechia",
        "Czechia": "Czechia",
    }

    return replacements.get(value, value)


qs["Country/Territory"] = qs["Country/Territory"].apply(
    standardize_country_name
)

the["Country"] = the["Country"].apply(
    standardize_country_name
)

print("Country names standardized successfully.")
```

```python
print("QS duplicate university names:",
      qs["Institution Name"].duplicated().sum())

print("THE duplicate university names:",
      the["Name"].duplicated().sum())

print("QS duplicate rows:",
      qs.duplicated().sum())

print("THE duplicate rows:",
      the.duplicated().sum())
```

```python
FINAL_FILE = OUTPUT_DIR / "university_cleaned_merged_data.csv"

final_df = pd.read_csv(
    FINAL_FILE,
    encoding="utf-8-sig"
)

print("Final integrated dataset shape:", final_df.shape)

display(final_df.head())
```

```python
print("Duplicate QS universities:",
      final_df["qs_Institution Name"].duplicated().sum())

print("Duplicate THE universities:",
      final_df["the_Name"].duplicated().sum())

print("Duplicate complete rows:",
      final_df.duplicated().sum())
```

```python
country_mismatch = (
    final_df["qs_Country/Territory"].astype(str)
    != final_df["the_Country"].astype(str)
)

print("Country mismatches:", country_mismatch.sum())
```

```python
core_fields = [
    "qs_Institution Name",
    "qs_Country/Territory",
    "qs_2026_rank",
    "qs_AR SCORE",
    "qs_ER SCORE",
    "qs_FSR SCORE",
    "qs_CPF SCORE",
    "qs_IFR SCORE",
    "qs_ISR SCORE",
    "qs_ISD SCORE",
    "qs_IRN SCORE",
    "qs_EO SCORE",
    "the_Name",
    "the_Country",
    "the_rank",
    "the_Student Population",
    "the_Students to Staff Ratio",
    "the_Overall Score",
    "the_Teaching",
    "the_Research Environment",
    "the_Research Quality",
    "the_Industry Impact",
    "the_International Outlook",
    "the_female_to_male_ratio_clean",
]

# Keep only fields that exist
core_fields = [
    col for col in core_fields
    if col in final_df.columns
]

total_cells = len(final_df) * len(core_fields)
missing_cells = final_df[core_fields].isna().sum().sum()

missing_percentage = (
    missing_cells / total_cells
) * 100

completeness = 100 - missing_percentage

print("CORE COMPLETENESS CHECK")
print("=" * 60)
print("Rows:", len(final_df))
print("Core fields:", len(core_fields))
print("Total core cells:", total_cells)
print("Missing core cells:", missing_cells)
print(f"Missing percentage: {missing_percentage:.2f}%")
print(f"Completeness: {completeness:.2f}%")

print("\nRequirement:")
print("Completeness > 95%")
print("Missing values < 2%")

if completeness > 95 and missing_percentage < 2:
    print("\nPASS")
else:
    print("\nFAIL")
```

```python
missing_by_field = (
    final_df[core_fields]
    .isna()
    .sum()
    .sort_values(ascending=False)
)

display(
    missing_by_field.to_frame("Missing Values")
)
```

```python
print("MATCH TYPE SUMMARY")
display(final_df["match_type"].value_counts())

print("\nMATCH SIMILARITY")
display(final_df["match_similarity"].describe())
```

```python
checks = {
    "Rows > 0": len(final_df) > 0,
    "Duplicate QS universities": final_df["qs_Institution Name"].duplicated().sum() == 0,
    "Duplicate THE universities": final_df["the_Name"].duplicated().sum() == 0,
    "Country mismatches": country_mismatch.sum() == 0,
    "Missing QS names": final_df["qs_Institution Name"].isna().sum() == 0,
    "Missing THE names": final_df["the_Name"].isna().sum() == 0,
    "Completeness > 95%": completeness > 95,
    "Missing values < 2%": missing_percentage < 2,
}

display(
    pd.DataFrame(
        {"Check": checks.keys(), "Passed": checks.values()}
    )
)

if all(checks.values()):
    print("\nFINAL STATUS: PASS")
    print("Dataset is ready for Tableau/dashboard development.")
else:
    print("\nFINAL STATUS: REVIEW REQUIRED")
```
