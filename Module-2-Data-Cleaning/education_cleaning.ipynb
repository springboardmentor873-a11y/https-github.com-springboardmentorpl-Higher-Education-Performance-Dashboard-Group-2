```python
#Imports & config
import pandas as pd
import numpy as np
import os

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 150)

DATA_DIR = "."

QS_FILE  = "2026 QS World University Rankings.csv"
THE_FILE = "THE World University Rankings 2016-2026.csv"

print("Files present:", os.listdir(DATA_DIR))
```

```python
# Module 1 - Load raw datasets
qs_raw = pd.read_csv(os.path.join(DATA_DIR, QS_FILE))
the_raw = pd.read_csv(os.path.join(DATA_DIR, THE_FILE))

print("QS raw shape :", qs_raw.shape)
print("THE raw shape:", the_raw.shape)

print("\nQS columns:\n", qs_raw.columns.tolist())
print("\nTHE columns:\n", the_raw.columns.tolist())

qs_raw.head()
```

```python
# Module 1 - Quick look at THE (latest year only, for sanity check)
the_raw[the_raw['Year'] == 2026].head()
```

```python
# Module 1 - Standardize each source into a common schema
# We tag each row with its ranking body so both can be merged into one file.
# Cell 4: Module 1 - Standardize each source into a common schema
# NOTE: THE dataset spans 2016-2026, but we only want 2026 to match QS 2026.

the_raw_2026 = the_raw[the_raw['Year'] == 2026].copy()
print("THE rows before filtering:", the_raw.shape[0])
print("THE rows for 2026 only  :", the_raw_2026.shape[0])

qs_common = pd.DataFrame({
    'source'            : 'QS',
    'year'               : 2026,
    'rank_raw'           : qs_raw['2026 Rank'],
    'previous_rank_raw'  : qs_raw['Previous Rank'],
    'institution'        : qs_raw['Institution Name'],
    'country'            : qs_raw['Country/Territory'],
    'region'             : qs_raw['Region'],
    'overall_score_raw'  : qs_raw['Overall SCORE'],
    'academic_reputation': qs_raw['AR SCORE'],
    'employer_reputation': qs_raw['ER SCORE'],
    'faculty_student_ratio_score': qs_raw['FSR SCORE'],
    'citations_per_faculty': qs_raw['CPF SCORE'],
    'international_faculty_score': qs_raw['IFR SCORE'],
    'international_students_score': qs_raw['ISR SCORE'],
    'international_research_network': qs_raw['IRN SCORE'],
    'employment_outcomes': qs_raw['EO SCORE'],
    'sustainability_score': qs_raw['SUS SCORE'],
})

the_common = pd.DataFrame({
    'source'            : 'THE',
    'year'               : the_raw_2026['Year'],
    'rank_raw'           : the_raw_2026['Rank'],
    'previous_rank_raw'  : np.nan,
    'institution'        : the_raw_2026['Name'],
    'country'            : the_raw_2026['Country'],
    'region'             : np.nan,
    'overall_score_raw'  : the_raw_2026['Overall Score'],
    'academic_reputation': np.nan,
    'employer_reputation': np.nan,
    'faculty_student_ratio_score': the_raw_2026['Students to Staff Ratio'],
    'citations_per_faculty': the_raw_2026['Research Quality'],
    'international_faculty_score': np.nan,
    'international_students_score': the_raw_2026['International Students'],
    'international_research_network': the_raw_2026['International Outlook'],
    'employment_outcomes': the_raw_2026['Industry Impact'],
    'sustainability_score': np.nan,
})

# Merge ranking datasets into a common structure
university_raw = pd.concat([qs_common, the_common], ignore_index=True)
print("Merged shape:", university_raw.shape)
university_raw.head()
```

```python
# Module 1 - Completeness check (evaluation criterion: >95% complete)
completeness = 100 * (1 - university_raw.isna().mean())
print(completeness.round(2))

overall_completeness = 100 * (1 - university_raw.isna().mean().mean())
print(f"\nOverall dataset completeness: {overall_completeness:.2f}%")
```

```python
# Module 1 - Save deliverable
os.makedirs("data", exist_ok=True)
university_raw.to_csv("data/university_raw_data.csv", index=False)
print("Saved: data/university_raw_data.csv")
```

```python
# Module 2 -  Remove duplicates
before = len(university_raw)
university_clean = university_raw.drop_duplicates(subset=['source', 'year', 'institution']).copy()
after = len(university_clean)
print(f"Removed {before - after} duplicate rows")
```

```python
# Module 2 - Standardize university (institution) names
def clean_institution_name(name):
    if pd.isna(name):
        return name
    name = str(name).strip()
    name = " ".join(name.split())          # collapse repeated whitespace
    name = name.replace("’", "'")           # normalize curly apostrophes
    return name

university_clean['institution'] = university_clean['institution'].apply(clean_institution_name)
university_clean['institution'].head(10)
```

```python
# Module 2 - Standardize country names
# QS uses long-form ("United States of America"); THE uses short-form ("United States").
# We map both to one consistent set of names.

country_map = {
    "United States of America": "United States",
    "United Kingdom": "United Kingdom",
    "Hong Kong SAR": "Hong Kong",
    "Macau SAR": "Macau",
    "Russian Federation": "Russia",
    "South Korea": "South Korea",
    "Korea, South": "South Korea",
    "Czechia": "Czech Republic",
    "Viet Nam": "Vietnam",
    "Iran, Islamic Republic of": "Iran",
}

university_clean['country'] = university_clean['country'].str.strip()
university_clean['country'] = university_clean['country'].replace(country_map)

print("Unique countries after standardization:", university_clean['country'].nunique())
university_clean['country'].value_counts().head(10)
```

```python
# Module 2 - Normalize ranking metrics
# rank_raw contains plain numbers ("1", "45"), tie markers ("7="), 
# range bands ("701-710"), and open-ended bands ("1401+").
# We create a clean numeric rank for sorting/filtering in Tableau, 
# while keeping the original display string.

import re

def parse_rank(value):
    """Return a numeric rank usable for sorting. Uses band lower-bound for ranges."""
    if pd.isna(value):
        return np.nan
    s = str(value).strip()
    s = s.replace("=", "")          # drop tie markers like "7="
    s = s.replace("+", "")          # "1401+" -> "1401"
    if "-" in s:                    # range band "701-710" -> take lower bound 701
        s = s.split("-")[0]
    try:
        return float(s)
    except ValueError:
        return np.nan

university_clean['rank_display'] = university_clean['rank_raw'].astype(str)
university_clean['rank_numeric'] = university_clean['rank_raw'].apply(parse_rank)

# Normalize overall score: QS uses "-" for band rows with no published score
def parse_score(value):
    if pd.isna(value):
        return np.nan
    s = str(value).strip()
    if s in ("-", "", "N/A"):
        return np.nan
    try:
        return float(s)
    except ValueError:
        return np.nan

university_clean['overall_score'] = university_clean['overall_score_raw'].apply(parse_score)

# International Students: THE stores this as "26%" (text) -> convert to numeric 0-100
def parse_percent(value):
    if pd.isna(value):
        return np.nan
    s = str(value).strip().replace("%", "")
    try:
        return float(s)
    except ValueError:
        return np.nan

university_clean.loc[university_clean['source'] == 'THE', 'international_students_score'] = (
    university_clean.loc[university_clean['source'] == 'THE', 'international_students_score'].apply(parse_percent)
)

university_clean[['rank_raw', 'rank_display', 'rank_numeric', 'overall_score_raw', 'overall_score']].sample(10, random_state=1)
```

```python
# Module 2 - Drop now-redundant raw columns, reorder for Tableau
tableau_ready_cols = [
    'source', 'year', 'institution', 'country', 'region',
    'rank_numeric', 'rank_display', 'previous_rank_raw', 'overall_score',
    'academic_reputation', 'employer_reputation', 'faculty_student_ratio_score',
    'citations_per_faculty', 'international_faculty_score',
    'international_students_score', 'international_research_network',
    'employment_outcomes', 'sustainability_score'
]

university_cleaned = university_clean[tableau_ready_cols].rename(columns={
    'previous_rank_raw': 'previous_rank'
})

university_cleaned.head()
```

```python
#Module 2 -  Missing value check (evaluation target: <2% missing overall)
missing_pct = university_cleaned.isna().mean() * 100
print(missing_pct.round(2))
print(f"\nOverall missing: {missing_pct.mean():.2f}%")
```

```python
# Module 2 -  Save deliverable
university_cleaned.to_csv("data/university_cleaned.csv", index=False)
print("Saved: data/university_cleaned.csv")
print("Final shape:", university_cleaned.shape)
```
