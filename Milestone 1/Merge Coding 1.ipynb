```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("/content/Merged_Higher_Education_Dashboard_50.xlsx")

df.head()
```

```python
print(df.columns.tolist())
```

```python
max_rank = df["QS_Rank"].max()
min_rank = df["QS_Rank"].min()

df["Global_Ranking_Score"] = (
    (max_rank - df["QS_Rank"]) /
    (max_rank - min_rank)
) * 100

df["Global_Ranking_Score"] = df["Global_Ranking_Score"].round(2)
```

```python
df["Research_Impact_Score"] = (
    df["Research"] * 0.6 +
    df["Citations"] * 0.4
).round(2)
```

```python
df["Faculty_to_Student_Ratio"] = (
    df["Faculty_Student_Ratio"]
).round(2)
```

```python
df["International_Student_Percentage"] = (
    df["International_Students"]
).round(2)
```

```python
df["Academic_Reputation_Score"] = (
    df["Academic_Reputation"]
).round(2)
```

```python
df["Academic_Reputation_Score"] = (
    df["Academic_Reputation"]
).round(2)
```

```python
df["Research_Productivity_Index"] = (
    (df["Research"] + df["Citations"] + df["Industry_Income"]) / 3
).round(2)
```

```python
df[
    [
        "University",
        "Global_Ranking_Score",
        "Research_Impact_Score",
        "Faculty_to_Student_Ratio",
        "International_Student_Percentage",
        "Academic_Reputation_Score",
        "Research_Productivity_Index"
    ]
].head()
```

```python
df.to_excel("university_final_dataset.xlsx", index=False)

print("KPI Engineering Completed Successfully!")
```
