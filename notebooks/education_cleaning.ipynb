```python
## Initial Data Quality Assessment

# Both source datasets were inspected for structure, duplicate records, and missing values.

# - QS contains 1,501 university records and 30 columns.
# - THE contains 2,191 university records and 14 columns.
# - No duplicate rows were identified in either source dataset.
# - Missing values are present in some ranking indicators and institutional attributes.
# - Missing values are retained where they represent unavailable source data rather than being replaced with artificial zero values.
```

```python
# Country-name standardization mapping

country_mapping = {
    "China (Mainland)": "China",
    "Hong Kong SAR, China": "Hong Kong",
    "Iran (Islamic Republic of)": "Iran",
    "Macao SAR, China": "Macao",
    "Republic of Korea": "South Korea",
    "Syrian Arab Republic": "Syria",
    "Türkiye": "Turkey",
    "United States of America": "United States",
    "Venezuela (Bolivarian Republic of)": "Venezuela",
    "Viet Nam": "Vietnam"
}
```

```python
qs["Standard Country"] = (
    qs["Country/Territory"]
    .replace(country_mapping)
)

the["Standard Country"] = (
    the["Country"]
    .replace(country_mapping)
)

print("Country standardization completed.")
```

```python
qs_countries = set(qs["Standard Country"].dropna().unique())
the_countries = set(the["Standard Country"].dropna().unique())

common_countries = qs_countries & the_countries

print("STANDARDIZED COUNTRY CHECK")
print("-" * 50)

print("QS unique countries:", len(qs_countries))
print("THE unique countries:", len(the_countries))
print("Common standardized countries:", len(common_countries))

print("\nQS-only countries:")
print(sorted(qs_countries - the_countries))

print("\nTHE-only countries:")
print(sorted(the_countries - qs_countries))
```

```python
## Country Standardization

# The QS and THE datasets use different naming conventions for several countries and territories.

# Examples include:

# - `United States of America` → `United States`
# - `China (Mainland)` → `China`
# - `Republic of Korea` → `South Korea`
# - `Türkiye` → `Turkey`
# - `Viet Nam` → `Vietnam`
# - `Iran (Islamic Republic of)` → `Iran`
# - `Syrian Arab Republic` → `Syria`
# - `Hong Kong SAR, China` → `Hong Kong`
# - `Macao SAR, China` → `Macao`

# A standardized country field was created while preserving the original source fields.

# Countries present in only one source were not artificially added to the other dataset or removed at this stage.
```

```python
def normalize_university_name(name):
    """
    Create a standardized matching key for university names.
    The original source names remain unchanged.
    """

    if pd.isna(name):
        return ""

    name = str(name).strip().lower()

    # Remove text inside parentheses
    name = re.sub(r"\([^)]*\)", "", name)

    # Remove leading "the"
    name = re.sub(r"^the\s+", "", name)

    # Standardize ampersand
    name = name.replace("&", "and")

    # Remove punctuation
    name = re.sub(r"[^\w\s]", " ", name)

    # Normalize multiple spaces
    name = re.sub(r"\s+", " ", name).strip()

    return name
```

```python
qs["University Match Name"] = (
    qs["Institution Name"]
    .apply(normalize_university_name)
)

the["University Match Name"] = (
    the["Name"]
    .apply(normalize_university_name)
)

print("University name normalization completed.")
```

```python
print("QS UNIVERSITY NAME NORMALIZATION")
print("-" * 80)

display(
    qs[
        ["Institution Name", "University Match Name"]
    ].head(20)
)

print("\nTHE UNIVERSITY NAME NORMALIZATION")
print("-" * 80)

display(
    the[
        ["Name", "University Match Name"]
    ].head(20)
)
```

```python
qs_names = set(qs["University Match Name"])
the_names = set(the["University Match Name"])

common_names = qs_names & the_names

print("UNIVERSITY NAME MATCHING")
print("-" * 50)

print("QS universities:", len(qs_names))
print("THE universities:", len(the_names))
print("Normalized exact matches:", len(common_names))

print(
    "QS match percentage:",
    round(len(common_names) / len(qs_names) * 100, 2),
    "%"
)

print(
    "THE match percentage:",
    round(len(common_names) / len(the_names) * 100, 2),
    "%"
)
```

```python
matched_data = pd.merge(
    qs,
    the,
    on=["University Match Name", "Standard Country"],
    how="inner",
    suffixes=("_QS", "_THE")
)

print("INNER JOIN RESULTS")
print("-" * 50)

print("QS universities:", len(qs))
print("THE universities:", len(the))
print("Matched universities:", len(matched_data))

print(
    "QS matched:",
    round(len(matched_data) / len(qs) * 100, 2),
    "%"
)

print(
    "THE matched:",
    round(len(matched_data) / len(the) * 100, 2),
    "%"
)
```

```python
## University Matching and Dataset Integration

# University names were normalized to create a matching key while preserving the original institution names.

# The normalization process:

# 1. Converts names to lowercase.
# 2. Removes text contained in parentheses.
# 3. Removes a leading `The`.
# 4. Standardizes ampersands.
# 5. Removes punctuation.
# 6. Normalizes whitespace.

# QS and THE records were then integrated using an **inner join** on:

# - `University Match Name`
# - `Standard Country`

# Only universities that could be matched across both datasets were retained for the comparative analysis dataset.

# The resulting dataset contains 870 matched universities.
```

```python
# --------------------------------------------------
# Create final Tableau-ready dataset
# --------------------------------------------------

final_data = pd.DataFrame({
    # University information
    "University Name": matched_data["Institution Name"],
    "Country": matched_data["Standard Country"],
    "Region": matched_data["Region"],

    # QS ranking information
    "QS Rank": matched_data["2026 Rank"],
    "QS Previous Rank": matched_data["Previous Rank"],
    "QS Overall Score": matched_data["Overall SCORE"],

    "QS Academic Reputation Score": matched_data["AR SCORE"],
    "QS Academic Reputation Rank": matched_data["AR RANK"],

    "QS Employer Reputation Score": matched_data["ER SCORE"],
    "QS Employer Reputation Rank": matched_data["ER RANK"],

    "QS Faculty Student Ratio Score": matched_data["FSR SCORE"],
    "QS Faculty Student Ratio Rank": matched_data["FSR RANK"],

    "QS Citations per Faculty Score": matched_data["CPF SCORE"],
    "QS Citations per Faculty Rank": matched_data["CPF RANK"],

    "QS International Faculty Ratio Score": matched_data["IFR SCORE"],
    "QS International Faculty Ratio Rank": matched_data["IFR RANK"],

    "QS International Student Ratio Score": matched_data["ISR SCORE"],
    "QS International Student Ratio Rank": matched_data["ISR RANK"],

    "QS International Student Diversity Score": matched_data["ISD SCORE"],
    "QS International Student Diversity Rank": matched_data["ISD RANK"],

    "QS International Research Network Score": matched_data["IRN SCORE"],
    "QS International Research Network Rank": matched_data["IRN RANK"],

    "QS Employment Outcomes Score": matched_data["EO SCORE"],
    "QS Employment Outcomes Rank": matched_data["EO RANK"],

    "QS Sustainability Score": matched_data["SUS SCORE"],
    "QS Sustainability Rank": matched_data["SUS RANK"],

    # THE ranking information
    "THE Rank": matched_data["Rank"],
    "THE Overall Score": matched_data["Overall Score"],

    "THE Student Population": matched_data["Student Population"],
    "THE Students to Staff Ratio": matched_data["Students to Staff Ratio"],
    "THE International Students": matched_data["International Students"],
    "THE Female to Male Ratio": matched_data["Female to Male Ratio"],

    "THE Teaching Score": matched_data["Teaching"],
    "THE Research Environment Score": matched_data["Research Environment"],
    "THE Research Quality Score": matched_data["Research Quality"],
    "THE Industry Impact Score": matched_data["Industry Impact"],
    "THE International Outlook Score": matched_data["International Outlook"],

    "Ranking Year": matched_data["Year"]
})

print("Tableau-ready dataset created.")
print("Rows:", len(final_data))
print("Columns:", len(final_data.columns))
```

```python
# --------------------------------------------------
# Remove duplicate university-country combinations
# --------------------------------------------------

duplicates_before = final_data.duplicated(
    subset=["University Name", "Country"]
).sum()

final_data = final_data.drop_duplicates(
    subset=["University Name", "Country"]
).reset_index(drop=True)

duplicates_after = final_data.duplicated(
    subset=["University Name", "Country"]
).sum()

print("Duplicate university-country records before:", duplicates_before)
print("Duplicate university-country records after:", duplicates_after)
print("Final rows:", len(final_data))
```

```python
# --------------------------------------------------
# Sort final dataset by QS ranking
# --------------------------------------------------

final_data = final_data.sort_values(
    by="QS Rank",
    ascending=True
).reset_index(drop=True)

display(
    final_data[
        [
            "University Name",
            "Country",
            "QS Rank",
            "THE Rank",
            "QS Overall Score",
            "THE Overall Score"
        ]
    ].head(10)
)
```

```python
# --------------------------------------------------
# Final data-quality validation
# --------------------------------------------------

missing_values = final_data.isna().sum()

missing_values = (
    missing_values[missing_values > 0]
    .sort_values(ascending=False)
)

total_cells = (
    final_data.shape[0] *
    final_data.shape[1]
)

missing_cells = final_data.isna().sum().sum()

missing_percentage = (
    missing_cells / total_cells
) * 100

print("FINAL DATA QUALITY VALIDATION")
print("-" * 50)

print("Rows:", len(final_data))
print("Columns:", len(final_data.columns))
print("Duplicate rows:", final_data.duplicated().sum())

print("\nColumns containing missing values:")
display(missing_values)

print(
    "\nOverall missing-value percentage:",
    round(missing_percentage, 2),
    "%"
)
```

```python
print("FINAL DATASET COLUMNS")
print("-" * 50)

for number, column in enumerate(final_data.columns, start=1):
    print(f"{number}. {column}")
```

```python
# --------------------------------------------------
# Export final cleaned dataset
# --------------------------------------------------

cleaned_file = OUTPUT_DIR / "university_cleaned.csv"

final_data.to_csv(
    cleaned_file,
    index=False
)

print("Cleaned dataset exported successfully.")
print("File:", cleaned_file)
```
